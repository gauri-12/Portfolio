```python
#Function definition for completeness checks

def complete_check(col_name):
    #check for null or missing values
    null_check = loans_df[col_name].isna()

    #calculate the ones that defy this rule
    check_fail_flag = null_check.sum()

    #Calculate the DQ score
    check_score = (1-(check_fail_flag/len(loans_df)))*100
    print("Percentage of rows where the", col_name, "is complete is:",check_score,"%")
```

```python
# Function for DQ Score calculation
DQ_Scores = {}
DQ_Failures = {}

def dq_score(flagged_rows, rule_name, denominator):

    # Handle the case where the denominator might be zero (e.g., no 'Fully Paid' loans)
    if denominator == 0:
        score = 100.0
        fail_count = 0
    else:
        fail_count = flagged_rows.sum()
        score = (1 - (fail_count / denominator)) * 100
    
    # Store the results
    DQ_Scores[rule_name] = round(score, 4)
    DQ_Failures[rule_name] = fail_count

    print(f"Percentage of rows where the check for {rule_name} is satisfied is: {score:.4f}%")
```

### addr_state
This variable has 4 DQ check rules defined. The below code blocks are operationalization of each of these 4 DQ rules.

#### Rule 1: Validity check
##### *The state should be represented as a 2 character value*

```python
#DQ rule code
length_check = loans_df['addr_state'].astype(str).str.len() != 2

#Calculate the DQ score
dq_score(length_check, 'State validity check', denominator=len(loans_df))
```

#### Rule 2: Completeness check
##### *No NULL values present*

```python
#DQ rule code
length_check = loans_df['addr_state'].isna()

#Calculate the DQ score
dq_score(length_check, 'State completeness check', denominator=len(loans_df))
```

#### Rule 3: Accuracy check
##### *The addr_state value must match a current, official 2-character USPS state code registry.*

```python
USPS_OFFICIAL_CODES = [
    'AL', 'AK', 'AZ', 'AR', 'CA', 'CO', 'CT', 'DE', 'DC', 'FL', 'GA', 
    'HI', 'ID', 'IL', 'IN', 'IA', 'KS', 'KY', 'LA', 'ME', 'MD', 'MA', 
    'MI', 'MN', 'MS', 'MO', 'MT', 'NE', 'NV', 'NH', 'NJ', 'NM', 'NY', 
    'NC', 'ND', 'OH', 'OK', 'OR', 'PA', 'PR', 'RI', 'SC', 'SD', 'TN', 
    'TX', 'UT', 'VT', 'VA', 'WA', 'WV', 'WI', 'WY'
]

#obtained from https://about.usps.com/who/profile/history/state-abbreviations.htm


#DQ rule code
accuracy_check = ~loans_df['addr_state'].isin(USPS_OFFICIAL_CODES)

#Calculate the DQ score
dq_score(accuracy_check, 'State accuracy check', denominator=len(loans_df))
```

### grade
This variable has 3 DQ check rules defined. The below code blocks are operationalization of each of these DQ rules.

#### Rule 1: Validity check
##### *The grade values should conform to the accepted domain values*

```python
accepted_grades = ['A', 'B', 'C', 'D', 'E', 'F', 'G']

#DQ rule code
domain_check = ~loans_df['grade'].isin(accepted_grades)

#Calculate the DQ score
dq_score(domain_check, 'Grade validity check', denominator=len(loans_df))
```

#### Rule 2: Completeness check
##### *No NULL values present*

```python
#DQ rule code
complete_check = loans_df['grade'].isna()

#Calculate the DQ score
dq_score(complete_check, 'Grade completeness check', denominator=len(loans_df))
```

#### Rule 3: Accuracy check
##### *The month-over-month distribution of the grades should not change by more than 5% (positive or negative)*

```python
# 1. Convert issue_d into a datetime format
loans_df['issue_d'] = pd.to_datetime(loans_df['issue_d'], format='%b-%Y')

# 2. Group by Month and Grade, then count occurrences
grade_counts_by_month = loans_df.groupby(['issue_d', 'grade']).size().reset_index(name='grade_count')


# 3. Calculate the Total Loans issued in each month
total_loans_by_month = loans_df.groupby('issue_d').size().reset_index(name='total_monthly_loans')


# 4. Merge the two DataFrames to align counts with totals
grade_distribution_df = pd.merge(
    grade_counts_by_month, 
    total_loans_by_month, 
    on='issue_d', 
    how='left'
)


# 5. Calculate the Final Distribution
grade_distribution_df['grade_percentage'] = (
    grade_distribution_df['grade_count'] / grade_distribution_df['total_monthly_loans']
)

#6. Sort the dataframe by the grade and the issue date to calculate MoM change
grade_distribution_df = grade_distribution_df.sort_values(by=['grade', 'issue_d'])

#7. Calculate the previous month's percentage
grade_distribution_df['previous_month_pct'] = grade_distribution_df.groupby('grade')['grade_percentage'].shift(1)

# 8. Calculate the Absolute Drift
grade_distribution_df['absolute_drift'] = abs(
    grade_distribution_df['grade_percentage'] - grade_distribution_df['previous_month_pct']
)
```

```python
# Calculate MoM change in grade value
MoM_change = grade_distribution_df['absolute_drift'] > 0.05

#Flag the ones that defy this rule
check3_fail_flag = MoM_change.sum()

dq_score(check3_fail_flag, 'Grade accuracy check', denominator=grade_distribution_df['absolute_drift'].count())
```

### loan_amnt

```python
#check the range of the loan amount field
range_check = (loans_df['loan_amnt'] < 500) | (loans_df['loan_amnt']>40000)

#Calculate the DQ score
dq_score(range_check, 'Loan Amount validity check', denominator=len(loans_df))
```
